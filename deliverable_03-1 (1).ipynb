## Task 1: Update the Character class

```python
from game_simulator_v1 import Character
```

```python
# Create two characters with different HP and attack damage
hero = Character("Hero", hp_max=100, damage=10)
enemy = Character("Enemy", hp_max=80, damage=15)

# Test the hero's attack against the enemy
hero.attack(enemy)
print("Enemy HP after attack:", enemy.hp)

# Test that the hero loses 30 HP
hero.take_damage(30)
print("Hero HP after damage:", hero.hp)

# Test that the hero regains 20 HP
hero.heal(20)
print("Hero HP after healing:", hero.hp)

# Check if the hero is still alive
print("Hero is alive:", hero.is_alive)

# Test that the enemy's HP does not go under 0
enemy.take_damage(100)
print("Enemy HP:", enemy.hp)
print("Enemy is alive:", enemy.is_alive)
```

We also changed `attack()` so it uses `take_damage()` instead of changing the enemy's HP directly. Finally, we added `is_alive` to check whether a character is still alive. The tests show that everything works as expected.

## Task 5: Running battles

```python
from game_simulator_v1 import Battle, Warrior, Mage

hero = Warrior("Hero")
enemy = Mage("Enemy")

battle = Battle(hero, enemy)

battle.simulate_battle()

print("Hero HP:", hero.hp)
print("Enemy HP:", enemy.hp)
print("Hero is alive:", hero.is_alive)
print("Enemy is alive:", enemy.is_alive)
```

In Task 5, we created a `Battle` class to simulate a fight between two characters. The `play_round()` method handles one round, where the hero attacks first and the enemy attacks back if it is still alive. The `simulate_battle()` method uses a `while` loop to repeat the rounds until one of the characters runs out of HP. The test shows that the battle works as expected and ends when one character is defeated.
